```python
import pandas as pd
from pathlib import Path
```

```python
# Set up the correct path to the raw folder
project_folder = Path.cwd().parent
raw_folder = project_folder / "data" / "raw"
```

```python
# Get the list of Excel files ONLY in 'raw'
excel_files = list(raw_folder.glob("*.xlsx"))
```

print(f"\nFound {len(excel_files)} Excel files in 'raw':\n")
for file in excel_files:
    print(file)

```python
# Read each file into a DataFrame out of the 'raw' folder and combine them into a single one
if excel_files:
    raw_data = pd.concat([pd.read_excel(file) for file in excel_files], ignore_index=True)
    print(f"Successfully combined {len(excel_files)} files! Shape: {raw_data.shape}")
else:
    print("No Excel files found in the 'raw' folder.")
```

```python
print(raw_data.shape)
# 504047, 250
```

```python
print(raw_data.head())
```

```python
#Remove leading and trailing whitespace from column names.
raw_data.columns = raw_data.columns.str.strip()
```

```python
selected_columns = [
    "Participant name",
    "Presented Media name",
    "Timeline name",
    "Recording timestamp  ms",
    "time_diff",
    "Gaze event duration  ms",
    "Validity left",
    "Validity right",
    "Eye movement type",
    "Gaze point X  DACS px",
    "Gaze point Y  DACS px",
    "AOIs"
]
raw_data_selected_01 = raw_data[selected_columns].copy()
```

```python
raw_data_selected_01.shape
```

```python
from pathlib import Path

# Save the transformed dataset to the interim data directory
# to preserve an intermediate analysis-ready version for subsequent steps.
interim_folder = project_folder / "data" / "interim"

csv_file_path = interim_folder / "raw_data_selected_01.csv"
raw_data_selected_01.to_csv(csv_file_path, index=False)
```
