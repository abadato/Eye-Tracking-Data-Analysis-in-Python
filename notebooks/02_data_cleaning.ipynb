```python
# Filter the dataset to retain valid fixation events with available participant
# and stimulus identifiers, excluding other eye-tracking events and incomplete records.
# This reduces noise and prepares the data for subsequent analysis.
raw_data_filtered= raw_data_selected_01[
    (raw_data_selected_01["Eye movement type"] == "Fixation") &
    (raw_data_selected_01["Participant name"].notna()) &
    (raw_data_selected_01["Presented Media name"].notna())
].copy()

print("Rows after filtering:", len(raw_data_filtered))

# The dataset was reduced from 504,047 to 385,898 rows,
# excluding approximately 23.44% of the original records.
```

```python
# To remove the exterm shorter and longer fixatoins
results_filtered = raw_data_filtered[
    (raw_data_filtered["Gaze event duration  ms"] >= 80) &
    (raw_data_filtered
    ["Gaze event duration  ms"] <= 2000)
].copy()
# Create an independent copy of the filtered dataset for subsequent processing.
```

```python
removed_percentage = 100 - retained_percentage

print(f"{removed_percentage:.2f}% of the data were removed.")

# 3.73% of the data were removed - the extrem long and shorter fixations.
```

```python
total_rows = len(raw_data_filtered)

short_fixations = (
    raw_data_filtered["Gaze event duration  ms"] < 80
).sum()

long_fixations = (
    raw_data_filtered["Gaze event duration  ms"] > 2000
).sum()

missing_duration = (
    raw_data_filtered["Gaze event duration  ms"].isna()
).sum()

results_filtered = raw_data_filtered[
    (raw_data_filtered["Gaze event duration  ms"] >= 80) &
    (raw_data_filtered  ["Gaze event duration  ms"] <= 2000)
].copy()

retained_rows = len(results_filtered)
removed_rows = total_rows - retained_rows

print(f"Original rows:       {total_rows:,}")
print(f"Short fixations:     {short_fixations:,}")
print(f"Long fixations:      {long_fixations:,}")
print(f"Missing duration:    {missing_duration:,}")
print(f"Rows retained:       {retained_rows:,}")
print(f"Rows removed:        {removed_rows:,}")
print(f"Retained:             {retained_rows / total_rows * 100:.2f}%")
print(f"Removed:              {removed_rows / total_rows * 100:.2f}%")
```

```python
print(results_filtered.columns.tolist())
```

```python
results_binned = results_filtered.copy() #  Parentheses need to copy as the DataFrame.
```

```python
# Calculate the onset of each stimulus/trial
results_binned["item_onset_time"] = (
    results_binned
    .groupby(["Participant name", "Presented Media name"])["Recording timestamp  ms"]
    .transform("min")
)

# Calculate time relative to stimulus onset
results_binned["ms"] = (
    results_binned["Recording timestamp  ms"]
    - results_binned["item_onset_time"]
)

print(results_binned[
    [
        "Participant name",
        "Presented Media name",
        "Recording timestamp  ms",
        "item_onset_time",
        "ms"
    ]
].head(10))
```

```python
#Create the 20-ms bins
BIN_DURATION = 20

results_binned["bin20"] = (
    BIN_DURATION *
    (results_binned["ms"] / BIN_DURATION).round()
)

print(results_binned[
    ["ms", "bin20"]
].head(20))
```

```python
#Rename the stimulus column
results_binned = results_binned.rename(
    columns={"Presented Media name": "items"}
)
```

```python
# Strip '.avi' extension and surrounding spaces from entries in the 'items' column
results_binned["items"] = (
    results_binned["items"]
    .astype(str)
    .str.replace(".avi", "", regex=False)
    .str.strip()
)
```

```python
print(results_binned.shape)
print(results_binned[
    [
        "Participant name",
        "items",
        "Recording timestamp  ms",
        "item_onset_time",
        "ms",
        "bin20"
    ]
].head(10))
```

```python
# Rename columns in results_binned using pandas syntax
results_binned = results_binned.rename(
    columns={
        "Recording timestamp  ms": "RecordingTimestamp",
        "Participant name": "ParticipantName",
        "Presented Media name": "MediaName",
        "Eye movement type": "EyeMovementType",
        "Gaze point X  DACS px": "GazePointX",
        "Gaze point Y  DACS px": "GazePointY",
        "AOIs": "AOI",
        "Timeline name": "Lists",
        "Gaze event duration  ms": "GazeEventDur_ms",
    }
)
```
