## 03 — Integrating Eye-Tracking Data and Stimulus Metadata

```python
### Merge observations with experimental stimulus information
```

```python
import pandas as pd
from pathlib import Path
```

```python
#Get the main project folder (moves up from 'notebooks' to root) ->Python_eye_tracking_2x2
project_folder = Path.cwd().parent
```

```python
# Construct the correct path to the interim folder
csv_file_path = project_folder / "data" / "interim" / "results_binned_02.csv"
```

```python
#Read the dataset into pandas
results_binned_02 = pd.read_csv(csv_file_path)
```

```python
results_binned = results_binned_02.copy()
```

```python
print(results_binned.shape)
```

```python
print(results_binned.head())
```

```python
# The metadata shoudl be loaded separately:
stimuli = pd.read_csv(
    project_folder / "data" / "metadata" / "NomDat_experimental_stimuli.csv", sep=";"
)
```

```python
stimuli.shape
stimuli.head()
```

```python
# to make the join key consistent the column name shout match
stimuli = stimuli.rename(
    columns={"Item": "items"}
)
```

```python
results_binned.head()
```

```python
stimuli.head()
```

```python
et_data_joined = results_binned.merge(
    stimuli,
    on="items",
    how="left"
)
```

```python
et_data_joined.head()
```

```python
# Create an independent copy of et_data_joined
et_data_joined_03 = et_data_joined.copy()
```

```python
from pathlib import Path

# Define the interim output path
interim_folder = project_folder / "data" / "interim"
# Define the file path
csv_file_path = interim_folder / "et_data_joined_03.csv"
# Export to CSV
et_data_joined_03.to_csv(csv_file_path, index=False)
```

All 80 unique stimulus IDs in the eye-tracking dataset matched the stimulus metadata. No unmatched eye-tracking stimulus IDs were detected. The merge preserved the total number of rows (371,510), indicating that the integration did not change the dataset's row count.
